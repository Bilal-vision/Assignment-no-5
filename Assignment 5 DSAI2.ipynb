Q1. Write a Python function count_vowels that accepts a string and returns the number of vowels a, e, i, o, u, A, E, I, O, U in that string.

- Example: 'Welcome'
- Output: 3 vowels

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    return sum(1 for character in text if character in vowels)

print(count_vowels("Welcome"))
```

Q2. Write a Python function display_student_info that accepts the following details about a student:

- name (Keyword argument): The name of the student.
- age (Keyword argument): The age of the student.

- school (Keyword argument): The name of the school the student attends.

Q3. Write a Python function display_prices that takes a product name and an arbitrary number of keyword arguments representing product prices. The function should display each product and its price.

Expected Output:

Product: Laptop

Amazon: $800

eBay: $750

Walmart: $780

```python
def display_prices(product_name, **prices):
    print(f"Product: {product_name}")
    for store, price in prices.items():
        print(f"{store}: ${price}")

display_prices("Laptop", Amazon=800, eBay=750, Walmart=780)
```

Q4. Write a Python function find_max that takes arbitrary number of positional arguments and returns the maximum number.

```python
def find_max(*numbers):
    if not numbers:
        raise ValueError("find_max requires at least one number")
    return max(numbers)
print(find_max(12, 5, 27, 9))
```

Q6.

Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested functions:

- calculate_area: This function should calculate the area of the rectangle.
- calculate_perimeter: This function should calculate the perimeter of the rectangle.
- The main function should return both the area and the perimeter of the rectangle.

Expected Output:

Area: 15

Perimeter: 16


```python
def calculate_rectangle(length, width):
    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    return calculate_area(), calculate_perimeter()

area, perimeter = calculate_rectangle(5, 3)
print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```

Q8. Write a nested function to calculate the total hotel bill based on the number of nights.
A hotel charges Rs. 5,000 per night. The customer must also pay 10% tax on the room cost

Expected Output:
Room Cost: 15000
Tax: 1500.0
Total Bill: 16500.0

```python
def calculate_hotel_bill(number_of_nights):
    rate_per_night = 5000

    def calculate_room_cost():
        return number_of_nights * rate_per_night

    room_cost = calculate_room_cost()
    tax = room_cost * 0.10
    total_bill = room_cost + tax
    return room_cost, tax, total_bill

room_cost, tax, total_bill = calculate_hotel_bill(3)
print(f"Room Cost: {room_cost}")
print(f"Tax: {tax}")
print(f"Total Bill: {total_bill}")
```

Q9.User Login Validation
A website checks a user's username and password before allowing access. Write a nested function to validate the login details

```python
def login(username, password, correct_username, correct_password):
    def validate_credentials():
        return username == correct_username and password == correct_password
        
    if validate_credentials():
        return "Login successful"
    return "Invalid username or password"
print(login("student", "pass123", "student", "pass123"))
```

Q10.

Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

- An inner function calculate_total() → multiplies price * quantity.
- An inner function apply_discount(total) → subtracts discount from total (but never below 0).

The outer function should return the final bill amount.

Expected Output:

- Final Bill = 600 PKR (no discount)
- Final Bill = 500 PKR (with discount)

```python
def shopping_bill(price, quantity, discount=0):
    def calculate_total():
        return price * quantity

    def apply_discount(total):
        return max(0, total - discount)

    return apply_discount(calculate_total())

print(f"Final Bill = {shopping_bill(200, 3)} PKR")      
print(f"Final Bill = {shopping_bill(200, 3, 100)} PKR")  
```
